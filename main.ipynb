```python
import cv2
import matplotlib.pyplot as plt

# Membaca foto
foto_bgr = cv2.imread("gambar/foto_saya.jpeg")

if foto_bgr is None:
    print("Foto tidak ditemukan!")
else:
    # Mengubah BGR menjadi RGB
    foto_asli = cv2.cvtColor(foto_bgr, cv2.COLOR_BGR2RGB)

    # Mengatur perubahan kecerahan
    cerah = 80
    gelap = -80

    # Menambahkan dan mengurangi kecerahan
    foto_cerah = cv2.convertScaleAbs(
        foto_asli, alpha=1, beta=cerah
    )

    foto_gelap = cv2.convertScaleAbs(
        foto_asli, alpha=1, beta=gelap
    )

    # Menampilkan perbandingan gambar
    plt.figure(figsize=(15, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(foto_asli)
    plt.title("Foto Asli")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(foto_cerah)
    plt.title("Foto Lebih Cerah")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(foto_gelap)
    plt.title("Foto Lebih Gelap")
    plt.axis("off")

    plt.tight_layout()
    plt.show()
    
```
